# Student Marks and Percentage calculator

###

```python
Name = "Mohammad Wasayullah"
Total_Marks = 560
Eng = 78
Urd = 77
Isl = 50 
Math = 93 
Ttq = 47
Phy = 67
Cs = 69 
Obtained_Marks = Eng+Isl+Urd+Math+Ttq+Phy+Cs
Percentage = (Obtained_Marks/Total_Marks)*100

if Percentage >=90:
    grade = "A+"
elif Percentage >= 80:
    grade = "A"
elif Percentage >= 70:
    grade = "B"
elif Percentage >= 50:
    grade = "C"
elif Percentage >= 40:
    grade = "D"
else:
    grade = "Fail"


```

#### OutPut

```python
print(f"""
================================================
| Name:{Name}                             
| Marks: {Obtained_Marks} Out of {Total_Marks} 
| Percentage:{Percentage:.2f} %     
| Grade : {grade}           
================================================
""")    
```
