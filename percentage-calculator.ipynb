# Student Marks and Percentage calculator

#### Result Calculation

```python
Name = input("Enter Your Name ")
Total_Marks = 560
Eng = eval(input("Enter Your marks in English"))
Math = eval(input("Enter Your marks in Mathematics")) 
Cs = eval(input("Enter Your marks in Computer Science")) 
Phy = eval(input("Enter Your marks in Physics"))
Urd = eval(input("Enter Your marks in Urdu"))
Isl = eval(input("Enter Your marks in Islamiat")) 
Ttq = eval(input("Enter Your marks in Tarjumat-ul-Quran"))
Obtained_Marks = Eng+Isl+Urd+Math+Ttq+Phy+Cs
Percentage = (Obtained_Marks/Total_Marks)*100

if Percentage >=90:
    grade = "A+"
elif Percentage >= 80:
    grade = "A"
elif Percentage >= 70:
    grade = "B"
elif Percentage >= 50:
    grade = "C"
elif Percentage >= 40:
    grade = "D"
else:
    grade = "Fail"


```

#### OutPut

```python
print(f"""
================================================
| MARKSHEET FOR ANNUAL EXAMINATION |

| Name:{Name}                             
| Marks: {Obtained_Marks} Out of {Total_Marks} 
| Percentage:{Percentage:.2f} %     
| Grade : {grade}     
      
================================================
""")    
```
